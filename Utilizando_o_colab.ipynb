```
import random

#criar uma lista com valores e ficar removendo algumas informações

lista_valores = [1,2,3,4,5,6]
sortear = random.choice(lista_valores)
print(sortear)
```

```
#Jogo - 1 Remover valores da lista

lista_pets = ["Meg","Duda","Pandora","Joy","Garoa","Mini", "Soneca", "Honda","Lucky","Bingo"]

sortear = random.choice(lista_pets)
print(sortear)
print(f"A lista tem os elementos:  {lista_pets}")

posicao = int(input('informe a posição do valor sorteado: '))
print(f'O valor removido foi {lista_pets[posicao]}')
lista_pets.pop(posicao)

print(lista_pets)
```
